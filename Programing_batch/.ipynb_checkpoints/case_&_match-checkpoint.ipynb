```python
#An electricity company categorizes consumption as:

#upto 100 units: "low consumption"
#101 to 301 units : "normal consuption"
#310 to 500 units : "High Consumption"
#above 500 : "very high Consumption""

c = int(input("Enter the units:"))

if c < 100:
    print("Low Consumption")
elif 101 <= c <= 301:
    print("Normal Consumption")
elif 310 <= c <= 500:
    print("High Consumption")
else:
    print("Very High Consumption")
    
```

```python
#nested if

#A student can pass an exam only if the student has 75% or more attendance.
#If the attendance requirement is satisfied, check the marks. If marks are 
# 40 or more, print "Student Passed", Otherwise print "Student failed Due to Low marks". If attendance is below 75%, print
# "not Eligible for Exam"
```

```python
attendance = int(input("Enter the Attendance Percentage: "))
marks = int(input("Enter marks:"))

if attendance >= 75:
    if marks >= 40:
        print("Student Passed")
    else:
        print("Student failed due to low marks")
else:
    print("Not eligible for exam")
```

```python
attendance = int(input("Enter the Attendance Percentage: "))
marks = int(input("Enter marks: "))

match attendance, marks:
    case attendance, marks if attendance >= 75:
        match marks:
            case marks if marks >= 40:
                print("student passed")
            case _:
                print("Student failed due to low marks")
    case _:
        print("not eligible for exam")
```

```python
#An ATM allows a transaction only when the entered PIN is correct. If the PIN is correct,
#check the withdraw1 amoun. If the amount is less than or equal to 100000, print "Withdraw1\
#Successful", othewise print "Withdraw1 Successful", othewise print "Daily Limit Exceeded". 
#If the pin is incorrect, print "Incorrect Pin".

pin = int(input("Enter the PIN:"))
amount = int(input("Enter thw amount:"))

if pin == 4321:
    if amount <= 10000:
        print("Withdrawl Successful")
    else:
        print("Daily Limit Exceeded")

else:
    print("Incorrect PIN")
    
```

```python
#wap to print fibonacci series of n numbers

n = int(input("Enter number of terms : "))
first, second = 0, 1
i = 1

while i <= n:
    print(first, end = " ")
    first, second = second, first + second
    i += 1
    
```

```python
#wap to check whether a number is fibonacci number / fibonacci series number or not

n = int(input("Enter number of terms : "))
a = int(input("Enter the number for checking:"))

first, second = 0, 1
i = 1

while i <= n:
    #print(first, end = " ")
    first, second = second, first + second
    i += 1

if a == first:
    print(f"matched :{first}")
else:
    print(f"Not a fivonacci series number")
    
```

```python
number = int(input("Enter a number : "))
first, second = 0, 1

while first < number:
    first, second = second, first + second
if first == number:
    print(f"{number} is fibonacci number")
else:
    print(f"{number} is not fibonacci number")
```

```python
#wap to convert an iteger number into a roman number

#inp = 28
#out = XXVIII
number = int(input("Enter a number: "))
roman_numbers = {1 : "I", 2 : "IV", 5 : "V", 9 : "IX", 10 : "X", 40 : "XL", 50 : "L", 90 : "XC", 100 : "C", 400 : "CD", 500 : "D", 900 : "CM", 1000 : "M"}
result = ""
keys = list (roman_numbers.keys())
i = len(keys) - 1

while number != 0:
    if number >= keys[i]:
        result += roman_numbers[keys[i]]
        number -= keys[i]
    else:
        i -= 1

print(result)
```

```python
# WAP to convert Roman number into integer number

# Input = MCMLXXVI
# Output = 1976

roman = input("Enter a ROMAN number: ")

roman_numbers = {
    "I": 1,
    "IV": 4,
    "V": 5,
    "IX": 9,
    "X": 10,
    "XL": 40,
    "L": 50,
    "XC": 90,
    "C": 100,
    "CD": 400,
    "D": 500,
    "CM": 900,
    "M": 1000
}

result = 0
i = 0

while i < len(roman):

    if i + 1 < len(roman) and roman[i:i+2] in roman_numbers:
        result += roman_numbers[roman[i:i+2]]
        i += 2

    else:
        result += roman_numbers[roman[i]]
        i += 1

print(result)
```

```python
#wap to convert roman nmber into integer number

#inp = MCMLXXIX
#OUT = 1976
roman = input("Enter a ROMAN numbers: ")
roman_numbers = {1 : "I", 2 : "IV", 5 : "V", 9 : "IX", 10 : "X", 40 : "XL", 50 : "L", 90 : "XC", 100 : "C", 400 : "CD", 500 : "D", 900 : "CM", 1000 : "M"}

keys = list(roman_numbers.keys())
i = len(keys) - 1
valid = ""

j = 0
while j < len(keys):
    valid += roman_numbers[keys[j]]
    j += 1

integer_number = 0

while roman != "":
    s = roman_numbers[keys[i]]
    if roman[0] not in valid:
        print("Invalid roman number")
        roman = ""
    elif roman[:len(s)] == s:
        integer_number += keys[i]
        roman = roman[len(s):]
        i = len(keys) - 1
    else:
        i -= 1
if valid != "":
    print(integer_number)
```
