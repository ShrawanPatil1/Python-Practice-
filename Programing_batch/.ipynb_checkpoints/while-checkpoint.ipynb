```python
#Binary to Hexadecimal
#inp = 1101
#out = D

binary = int(input("Enter binary number :"))
decimal = 0 
place, valid = 1, 1

while binary != 0:
    digit = binary % 10

    if digit == 0 or digit == 1:
        decimal += digit * place 
        place *= 2
        binary //= 10
    else:
        valid = 0
        binary = 0

if valid == 1:
    number = ""
    place = 1

    while decimal != 0:
        remainder = decimal % 16

        if remainder < 10:
            number = str(remainder) + number
        else:
            number = chr(remainder + 55) + number
        decimal //= 16
    print(number)

else:
    print("invalid binary number")
```

```python
# Octal to Hexadecimal
# inp = 17
# out = F

# inp = 2001
# out = 401

octal = int(input("Enter octal number: "))

decimal = 0
place = 1
valid = 1

# Octal to Decimal
while octal != 0:
    digit = octal % 10

    if digit >= 0 and digit <= 7:
        decimal += digit * place
        place *= 8
        octal //= 10
    else:
        valid = 0
        octal = 0

# Decimal to Hexadecimal
if valid == 1:

    number = ""

    if decimal == 0:
        number = "0"

    while decimal != 0:
        remainder = decimal % 16

        if remainder < 10:
            number = str(remainder) + number
        else:
            number = chr(remainder + 55) + number

        decimal //= 16

    print(number)

else:
    print("Invalid octal number")
```

```python
# Octal to Hexadecimal
# inp = 17
# out = F

# inp = 2001
# out = 401

octal = int(input("Enter octal number : "))
decimal = 0
place, valid = 1, 1

while octal != 0:
    digit = octal % 10

    if 0 <= digit <= 7:
        decimal += digit * place
        place *= 8
        octal //= 10
    else:
        valid = 0
        octal = 0

if valid == 1:
    number = ""
    place = 1

    while decimal != 0:
        remainder = decimal % 16

        if remainder < 10:
            number = str(remainder) + number
        else:
            number = chr(remainder + 55) + number
        decimal //= 16
    print(number)
else:
    print("Invalid Octal Number")
```

```python
#Hexadecimal to Binary 
#inp = D
#out = 1101

#inp = B5
#out = 10110101

hexadecimal = input("Enter Hexadecimal number :")
decimal, place, valid = 0, 1, 1

while hexadecimal != "":
    digit = hexadecimal[-1]

    if "0" <= digit <= "9":
        digit = int(digit)
        decimal += digit * place
        place *= 16
        hexadecimal = hexadecimal[:-1]

    elif "A" <= digit <= "F":
        digit = ord(digit) - 55
        decimal += digit * place 
        place *= 16
        hexadecimal = hexadecimal[:-1]
    else:
        valid, hexadecimal = 0, ""

if valid == 1:
    number, place = "", 1

    while decimal != 0:
        remainder = decimal % 2
        number = str(remainder) + number
        decimal //= 2
    print(number)

else:
    print("INvalid Hexadecimal number")
    
```

```python
# Hexadecimal to Octal
# inp = F
# out = 17

# inp = 401
# out = 2001

hexadecimal = input("Enter hexadecimal number : ").upper()

decimal = 0
place, valid = 1, 1
i = len(hexadecimal) - 1

while i >= 0:

    digit = hexadecimal[i] 

    if digit >= '0' and digit <= '9':
        value = ord(digit) - 48

    elif digit >= 'A' and digit <= 'F':
        value = ord(digit) - 55

    else:
        valid = 0
        i = -1
        continue

    decimal += value * place
    place *= 16
    i -= 1


if valid == 1:

    number = ""

    if decimal == 0:
        number = "0"

    while decimal != 0:

        remainder = decimal % 8
        number = str(remainder) + number
        decimal //= 8

    print(number)

else:
    print("Invalid Hexadecimal Number")
```

```python
# wap to remove duplicate digits from a given number and print the number with only unique digits in same order

#inp = 12324
#out = 1234

number = int(input("Enter a number :"))
temp_number = number
reverse = 0 

while temp_number != 0:
    digit = temp_number % 10
    reverse = reverse * 10 + digit
    temp_number //= 10
print('reverse is',reverse)
temp_number = reverse

seen_digits = {}
result = 0
place = 1

while temp_number != 0:
    digit = temp_number % 10
    if digit not in seen_digits:
        seen_digits[digit] = 1
        result += digit * place
        place *= 10
    temp_number //= 10
print('2nd loop result',result)
final_result = 0

while result != 0:
    digit = result % 10
    final_result = final_result * 10 + digit
    result //= 10

print('last loop result',final_result)
```

```python
#IMP PROGRAM IN TEST IT WILL COME
#wap to find  the most frequent digit from a number 
# inp = 1223332245
# out = 2

number = int(input("Enter a number :"))
digit_frequency = [0] * 10
temp_number = number

while temp_number != 0:
    digit = temp_number % 10
    digit_frequency[digit] += 1
    temp_number //= 10

highest_frequency = 0
most_frequent_digit = 0

temp_number = number

while temp_number != 0:
    digit = temp_number % 10
    if digit_frequency[digit] > highest_frequency:
        highest_frequency = digit_frequency[digit]
        most_frequent_digit = digit
    temp_number //= 10

print(most_frequent_digit)
```
