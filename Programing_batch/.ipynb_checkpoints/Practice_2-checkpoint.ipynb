```python
# Que

# Reverse a 3 digit number:

number = int(input("Enter the 3 digit number: "))

digit_3 = number % 10
digit_2 = (number // 10) % 10
digit_1 = number // 100

reverse = digit_3 * 100 + digit_2 * 10 + digit_1
print(f"The reverse of {number} string is: {reverse}")
```

```python
# wap to reverse a 4 digit nummber

number = int(input("Enter the 4 digit number: "))

digit_3 = number % 10
digit_2 = (number // 10) % 10
digit_1 = (number // 100) % 10
digit_0 = number // 1000

reverse = digit_3 * 1000 + digit_2 * 100 + digit_1 * 10 + digit_0
print(f"The reverse of {number} string is: {reverse}")
```

```python
# WAP to find the sum of digits of a 5 digit number without using loops or string operations.
#inp = 58321
#out = 19

number = int(input("Enter the 5 digit number:"))

digit_5 = number % 10
digit_4 = (number // 10) % 10
digit_3 = (number // 100) % 10
digit_2 = (number // 1000) % 10
digit_1 = number // 10000

summ = digit_5 + digit_4 + digit_3 + digit_2 + digit_1
print(f"The sum of {number} is : {summ}")
```

```python
#30.
#WAP to accept a 5 digit number and create the following output using ONLY arithmetic operators:
#1. Reverse of the number
#2. Sum of all digits
#3. Product of all digits
#4. Sum of squares of all digits
#5. Difference between the original number and reverse
#inp = 12345
#out =
#Reverse = 54321
#Digit Sum = 15
#Digit Product = 120
#Square Sum = 55
#Difference = -41976

number = int(input("Enter the 5 digit number:"))

digit_5 = number % 10
digit_4 = (number // 10) % 10
digit_3 = (number // 100) % 10
digit_2 = (number // 1000) % 10
digit_1 = number // 10000

reverse = digit_5 * 10000 + digit_4 * 1000 + digit_3 * 100 + digit_2 * 10 + digit_1
print(f"The reverse of the {number} is: {reverse}")

summ = digit_1 + digit_2 + digit_3 + digit_4 + digit_5  
print(f"The sum of all {number} is: {summ}")

product = digit_1 * digit_2 * digit_3 * digit_4 * digit_5
print(f"The product of all {number} is: {product}")

square = (digit_1 ** 2) + (digit_2 ** 2) + (digit_3 ** 2) + (digit_4 ** 2) + (digit_5 ** 2)
print(f"The square of all {number} is: {square}")

differnce = number - reverse
print(f"the reverse is {reverse} and number is {number} the differnce is : {differnce}")
```

```python
# 40.Best Student Award — Ultimate Challenge: A college wants to automatically identify its best
# student. The student must have attendance of at least 90%, Python marks of at least 85, must
# have completed at least 5 projects, and must have a discipline rating of "Excellent". Accept all
# four values and display "BEST STUDENT AWARD " only when every requirement is satisfied. 

attendance = int(input("Enter the attendance percent: "))
python = int(input("Enter the Python marks: "))
project = int(input("Enter the project count: "))
rating = input("Enter the Student Rating -Excellent | Not-: ")

if attendance >= 90 and python >= 85 and project >= 5 and rating == "Excellent":
    print("BEST STUDENT AWARD")
```

```python
# 39.Employee Promotion: A company considers an employee for promotion when their experience is
# at least 3 years, their performance rating is at least 8, and their attendance is at least 90%. Accept
# all three values and display "Promotion Eligible" only when all requirements are satisfied. 

exp = int(input("Enter the company experence: "))
performance = int(input("Enter the Performance rating "))
attendance = int(input("Enter the attendance :"))

if exp >= 3 and performance >= 8 and attendance >= 90:
    print( "Promotion Eligible")
```

```python
# 49. E-Commerce Delivery Challenge
# Accept cart value, membership type, and delivery distance.
# • If cart value is at least ₹5,000, delivery is free.
# • Otherwise, charge ₹100.
# • If distance is greater than 20 km, add ₹200 delivery charge.
# • If membership is "Premium", remove the ₹100 basic delivery charge.
# • Display the final delivery charge.
# Challenge: Carefully decide which charges should actually be applied.

cart = int(input("Enter the cart value:"))
m = input("Enter the membership type , Premium - or - NOt: ")
distance = int(input("Enter the delivery distance:"))
a = 0

if cart >= 5000:
    print("Delivery is free")
else:
    a = 100
if distance >= 20:
    a += 200
if m == "Premium":
    a -= 100
print(f"The final chagrge is: {a}")
```
